# Retail Sales Analysis

## Data Cleaning

### Business Objective

The objective of this phase is to prepare the dataset for exploratory analysis by correcting data types, handling missing values, validating data consistency, and creating new features that will support business insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

# Display numbers with two decimal places
pd.options.display.float_format = "{:.2f}".format
```

```python
# Load dataset
df = pd.read_csv("../data/raw/Superstore.csv")

# Display first rows
df.head()
```

```python
# Initial dataset information
df.info()
```

```python
# Convert date columns to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
# Verify data types
df[["Order Date", "Ship Date"]].dtypes
```

### Date Conversion

The **Order Date** and **Ship Date** columns were converted from text (`object`) to `datetime` format. This transformation enables time-based analyses such as monthly sales trends, yearly performance, and shipping time calculations.

## Missing Values

```python
# Check missing values
df.isnull().sum()
```

```python
# Display rows with missing Postal Code
df[df["Postal Code"].isnull()]
```

```python
# Remove rows with missing Postal Code
df = df.dropna(subset=["Postal Code"])

# Verify missing values
df.isnull().sum()
```

### Missing Values Handling

The dataset contained **11 missing values** in the **Postal Code** column, representing approximately **0.11%** of the total records.

Since Postal Code is not a critical variable for the business analysis and the proportion of missing values is negligible, the affected records were removed to maintain data quality without introducing artificial values.

## Duplicate Records

```python
# Check duplicate records
df.duplicated().sum()
```

```python
# Remove duplicates if any exist
df = df.drop_duplicates()

# Verify duplicates
df.duplicated().sum()
```

### Duplicate Records

No duplicate transactions were identified in the dataset. Therefore, no records were removed during this step.

## Data Consistency Validation

```python
# Check minimum values
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
# Negative sales
df[df["Sales"] < 0]
```

```python
# Quantity less than or equal to zero
df[df["Quantity"] <= 0]
```

```python
# Discounts outside expected range
df[(df["Discount"] < 0) | (df["Discount"] > 1)]
```

```python
# Negative profits
df[df["Profit"] < 0].head()
```

### Data Consistency Validation

The numerical variables were reviewed to identify potential inconsistencies.

No negative sales values were detected.

No quantities equal to or below zero were found.

All discount values fall within the expected range between **0** and **1**.

Negative profit values are present in the dataset. These observations are considered valid business cases rather than data quality issues, since discounts and operational costs may lead to unprofitable sales.

## Feature Engineering

```python
# Create Year column
df["Year"] = df["Order Date"].dt.year
```

```python
# Create Month column
df["Month"] = df["Order Date"].dt.month
```

```python
# Create Month Name column
df["Month Name"] = df["Order Date"].dt.month_name()
```

```python
# Create Quarter column
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
# Create Shipping Days column
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
# Display new columns
df.head()
```

### Feature Engineering

Several new variables were created to facilitate business analysis.

- **Year** enables yearly sales comparisons.
- **Month** supports monthly trend analysis.
- **Month Name** improves dashboard readability.
- **Quarter** allows quarterly performance evaluation.
- **Shipping Days** measures the time elapsed between order placement and shipment.

## Save Clean Dataset
